```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate some data
# Create 100 evenly spaced points between 0 and 10 for the X-axis
x = np.linspace(0, 10, 100)

# Calculate the sine of each X value for the Y-axis
y = np.sin(x)
```

```python
# 2. Create the plot
# Plot x and y, choose a color, and give it a label for the legend
plt.plot(x, y, color='blue', label='Sensor Wave')

# 3. Add titles and labels
plt.title("My Jupyter Plot")
plt.xlabel("Time (seconds)")
plt.ylabel("Sensor Value")
plt.grid(True)     # Adds a grid to the background
plt.legend()       # Displays the label we created in step 2
```
