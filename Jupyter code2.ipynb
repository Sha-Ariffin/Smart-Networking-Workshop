```python
import random
import json

# Loop 6 times (from 1 to 6)
for i in range(1, 7):
    # 1. Create simulated sensor data reading
    reading = {
        "reading_id": i,
        "temperature": round(random.uniform(20.0, 25.0), 2),
        "humidity": round(random.uniform(40.0, 60.0), 2)
    }
    
    # 2. Convert this single reading to a JSON string
    json_line = json.dumps(reading)
    
    # 3. Prints the output
    print(json_line)
```
