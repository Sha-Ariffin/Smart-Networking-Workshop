```python
#Import the library
from paho.mqtt import client as mqtt
import json
import random
import time
print ('ok')
```

```python
#Connect the broker
broker = "broker.hivemq.com"
port = 1883
topic = "mqtt-test"

client = mqtt.Client()

client.connect(broker, port)
print("Connected!")
```

```python
#Publish a single message
payload = {
    "temperature": 28.5,
    "humidity": 70,
    "distance": 35,
    "air_quality": 320
}

client.publish(topic, json.dumps(payload))
print("Message sent!")
```

```python
#Simulate a Pico W sensing sensor reading severy 2 seconds
while True:

    payload = {
        "temperature": round(random.uniform(25, 35), 1),
        "humidity": random.randint(60, 85),
        "distance": random.randint(10, 100),
        "air_quality": random.randint(200, 600)
    }

    client.publish(topic, json.dumps(payload))

    print(payload)

    time.sleep(2)
```
